```python
# STEP 1 — Import required libraries

import pandas as pd
import matplotlib.pyplot as plt

from google.colab import files
from sklearn.cluster import KMeans


# STEP 2 — Upload CSV file

# uploaded = files.upload()


# STEP 3 — Load dataset

df = pd.read_csv("fitness_cluster_data.csv")


# STEP 4 — Display dataset

print("Dataset:\n")
print(df)


# STEP 5 — Select required columns

X = df[['ExerciseHours', 'CaloriesBurned']]


# STEP 6 — Visualize the original data

plt.scatter(X['ExerciseHours'], X['CaloriesBurned'])

plt.xlabel("Exercise Hours")
plt.ylabel("Calories Burned")
plt.title("Customer Fitness Data")

plt.show()


# STEP 7 — Create K-Means model

model = KMeans(n_clusters=3)


# STEP 8 — Train the model

model.fit(X)


# STEP 9 — Get cluster labels

clusters = model.labels_

print("\nCluster Labels:\n")
print(clusters)


# STEP 10 — Add cluster labels to dataset

df['Cluster'] = clusters

print("\nUpdated Dataset:\n")
print(df)


# STEP 11 — Visualize clustered data

plt.scatter(df['ExerciseHours'], df['CaloriesBurned'], c=df['Cluster'])

plt.xlabel("Exercise Hours")
plt.ylabel("Calories Burned")

plt.title("K-Means Clustering")
plt.show()
```
