```python
# Scenario: Customer Segmentation Based on Shopping Behavior
# 🎯 Context
# A retail store wants to group its customers into segments based on their shopping patterns. They collected data on two features:
# Number of items purchased
# Total amount spent (₹ in thousands)
# They suspect there are two distinct customer groups:
# Budget shoppers (buy fewer items, spend less)
# Premium shoppers (buy more items, spend more)
# Items PurchasedﾠAmount Spent (₹k)
# 1ﾠ2
# 2ﾠ3
# 8ﾠ8
# 9ﾠ10
# 10ﾠ12
# 11ﾠ14
# Tasks for Students
# Load the dataset into Python using pandas.
# Apply K‑Means clustering with 2 clusters.
# Train the model to group customers.
# Visualize the clusters with a scatter plot.
# Interpret the results (Budget vs Premium shoppers).


# STEP 1 — Import required libraries

import pandas as pd
import matplotlib.pyplot as plt

from google.colab import files
from sklearn.cluster import KMeans


# STEP 2 — Upload CSV file

# uploaded = files.upload()


# STEP 3 — Load dataset

data = {
    'Numberofitemspurchased': [1,2,8,9,10,11],
    'Totalamountspent': [2000,3000,8000,10000,12000,14000]
}
df = pd.DataFrame(data)


# STEP 4 — Display dataset

print("Dataset:\n")
print(df)


# STEP 5 — Select required columns

X = df[['Numberofitemspurchased', 'Totalamountspent']]


# STEP 6 — Visualize the original data

plt.scatter(X['Numberofitemspurchased'], X['Totalamountspent'])

plt.xlabel("Number of items purchased")
plt.ylabel("Total amount spent")
plt.title("Shopping Behavior")

plt.show()


# STEP 7 — Create K-Means model

model = KMeans(n_clusters=2)


# STEP 8 — Train the model

model.fit(X)


# STEP 9 — Get cluster labels

clusters = model.labels_

print("\nCluster Labels:\n")
print(clusters)


# STEP 10 — Add cluster labels to dataset

df['Cluster'] = clusters

print("\nUpdated Dataset:\n")
print(df)


# STEP 11 — Visualize clustered data

plt.scatter(df['Numberofitemspurchased'], df['Totalamountspent'], c=df['Cluster'])

plt.xlabel("Number of items purchased")
plt.ylabel("Total amount spent")

plt.title("K-Means Clustering")
plt.show()
```
