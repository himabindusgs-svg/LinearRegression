```python
# Email Spam Detector (NLP)
# SCENARIO
# A company receives thousands of emails daily.
# The company wants AI to automatically identify:
# Spam Emails
# Non-Spam Emails
# CONCEPTS USED
# ✅ NLP
# ✅ Tokenization
# ✅ Stopwords Removal
# ✅ Sentiment/Text Classification
# ✅ Prediction Workflow
# PROJECT FLOW
# Email Text
# ↓
# Text Cleaning
# ↓
# Feature Extraction
# ↓
# Spam Prediction
# BEGINNER EXAMPLES
# EmailﾠPrediction
# “You won a free lottery”ﾠSpam
# “Meeting scheduled tomorrow”ﾠNot Spam


# STEP 1 — Import Libraries

import nltk
import spacy

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.sentiment import SentimentIntensityAnalyzer


# STEP 2 — Download Required NLTK Data

nltk.download('punkt_tab')
nltk.download('stopwords')


spam_text = 'You won a free lottery'

non_spam = 'Meeting scheduled tomorrow'

words_spam = word_tokenize(spam_text.lower())
words_non_spam = word_tokenize(non_spam.lower())

stop_words = set(stopwords.words('english'))

filtered_words_spam = []
filtered_words_nonspam = []

for words_spam in stop_words:
    if words_spam.lower() not in stop_words:
        filtered_words_spam.append(words_spam)

for words_non_spam in stop_words:
    if words_non_spam.lower() not in stop_words:
        filtered_words_nonspam.append(words_non_spam)

print("Spam words")
print(filtered_words_spam)
print("Non spam words")
print(filtered_words_nonspam)


# STEP 4 — Sentiment Analysis


nltk.download('vader_lexicon')

analyzer = SentimentIntensityAnalyzer()

# text = "Food was not okay. But it can be improved with more nutrition values."




# Get sentiment scores
scores_spam = analyzer.polarity_scores(spam_text)
scores_non_spam = analyzer.polarity_scores(non_spam)

# Print sentiment scores
print("Sentiment Scores for Spam:")
print(scores_spam)
print("Sentiment Scores for Non-Spam:")
print(scores_non_spam)

# Output results for. spam emails
print(f"Scores: {scores_spam}")
if scores_spam['compound'] >= 0.05:
    print("Overall Sentiment: Positive")
elif scores_spam['compound'] <= -0.05:
    print("Overall Sentiment: Negative")
else:
    print("Overall Sentiment: Neutral")

# Output results Non spam emails
print(f"Scores: {scores_non_spam}")
if scores_non_spam['compound'] >= 0.05:
    print("Overall Sentiment: Positive")
elif scores_non_spam['compound'] <= -0.05:
    print("Overall Sentiment: Negative")
else:
    print("Overall Sentiment: Neutral")
```
