```python
# STEP 1 — Import required libraries
import pandas as pd
import matplotlib.pyplot as plt
from google.colab import files
from sklearn.linear_model import LinearRegression

# STEP 2 — Upload CSV file
# uploaded = files.upload()

# STEP 3 — Load dataset
df = pd.read_csv("Student_Marks.csv")

# STEP 4 — Display dataset
print("Dataset:\n")
print(df)

# Prepare features (X) and target (y)
X = df[['StudyHours']]   # 2D DataFrame
y = df['Marks']          # 1D Series

# Train model
model = LinearRegression()
model.fit(X, y)

# Model parameters
print("\nSlope (m):", model.coef_[0])
print("Intercept (b):", model.intercept_)

# Predict marks for given study hours
y_pred = model.predict([[3]])   # Example: predict for 3 study hours
print("Predicted Marks for 3 study hours:", y_pred[0])
```
