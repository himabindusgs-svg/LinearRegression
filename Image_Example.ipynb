```python
# Scenario: Converting an Image to Grayscale
# 🎯 Context
# A student wants to learn how to use OpenCV for basic image manipulation. The simplest starting point is to load a color image and convert it into grayscale.
# 📂 Dataset
# Any sample image file (e.g., "sample.jpg") placed in the working directory.


import cv2
import matplotlib.pyplot as plt

from google.colab import files


#uploaded = files.upload()


image = cv2.imread("mountain.jpg")


image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)


plt.imshow(image_rgb)

plt.title("Scenary")

plt.axis("off")

plt.show()




gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)


plt.imshow(gray, cmap='gray')

plt.title("Grayscale Image")

plt.axis("off")
plt.show()
```
