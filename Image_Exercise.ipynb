```python
# PROJECT SCENARIO
# A supermarket wants AI to identify:
# apples,
# bananas,
# oranges
# from uploaded images automatically.
# The AI system should:
# load image dataset,
# preprocess images,
# train a model,
# classify uploaded fruit image.


# Import Libraries
import cv2
import matplotlib.pyplot as plt

from google.colab import files


import tensorflow as tf
from tensorflow.keras.datasets import cifar10
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Flatten, Dense




# CIFAR-10 Class Names
class_names = [
    'apples',
    'bananas',
    'oranges',
    'pineapple',
    'watermelon',
    'Mangos',
    'Strawberry',
    'Blueberry',
    'Kiwi',
    'Peach',
    'Pear',
    'Plum',
    'Pomegranate',
    'Avocado'
]


uploaded = files.upload()


image = cv2.imread("Mango.jpeg")


image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)


plt.imshow(image_rgb)

plt.title("Fruits")

plt.axis("off")

plt.show()


# Load Dataset
(X_train, y_train), (X_test, y_test) = cifar10.load_data()

# Normalize the image
X_train = X_train / 255.0
X_test = X_test / 255.0


# Create the model
model = Sequential()


# Flatten the layer
model.add(Flatten(input_shape=(32,32,3)))


# Hidden layer
model.add(Dense(128, activation='relu'))


# Output Layer
model.add(Dense(10, activation='softmax'))


# Compile model
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)


# Train the model
model.fit(X_train, y_train, epochs=100)


# STEP 10 — Predict Image

prediction = model.predict(X_test)


# Get prediction for first image
predicted_label = prediction[0].argmax()


print("Predicted Class:\n")

print(class_names[predicted_label])


print("Predicted Class:\n")

print(class_names[predicted_label])

```
