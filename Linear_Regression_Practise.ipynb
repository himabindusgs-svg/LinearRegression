```python
# Scenario: Predicting House Rent Based on Number of Rooms
# Context
# A property rental agency wants to estimate the monthly rent (₹) of apartments based on the number of rooms. They believe there’s a roughly linear relationship:
# Rent=m×Rooms+b

# Tasks
# Load the dataset into Python.
# Visualize with a scatter plot (Rooms vs Rent).
# Train a Linear Regression model using sklearn.
# Predict rent for a 3.5‑room apartment.
# Evaluate with R² score


import pandas as pd
import matplotlib.pyplot as plt

from google.colab import files
from sklearn.linear_model import LinearRegression


# STEP 2 — Upload CSV file

# uploaded = files.upload()


# STEP 3 — Load dataset

data = {
    'Rooms': [1, 2, 3, 4, 5],
    'Rent': [8, 12, 16, 20, 24]
}

df = pd.DataFrame(data)

print(df)


# STEP 5 — Visualize the data

plt.scatter(df['Rooms'], df['Rent'])

plt.xlabel("Rooms")
plt.ylabel("Rent")

plt.title("Rooms vs Rent")

plt.show()


# STEP 6 — Prepare input and output data

X = df[['Rooms']]   # Input
y = df['Rent']          # Output


# STEP 7 — Create Linear Regression model

model = LinearRegression()

# STEP 8 — Train the model

model.fit(X, y)


# STEP 9 — Predicted Fuel Efficiency for 10 litres

prediction = model.predict([[3.5]])

print("\nPredicting House Rent Based on Number of Rooms:\n")
print(prediction)


# STEP 10 — Draw Regression Line

plt.scatter(df['Rooms'], df['Rent'])

plt.plot(df['Rooms'], model.predict(X))

plt.xlabel("Rooms")
plt.ylabel("Rent")

plt.title("Linear Regression Line")

plt.show()










```
