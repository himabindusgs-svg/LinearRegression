```python
#Predicting Customer Purchase Based on Discounts
# 🎯 Context
#A retail store wants to predict whether a customer will buy a product (Yes/No) based on two factors:
# Discount Percentage offered
# Customer Income Level
# The idea is that higher discounts and higher incomes increase the likelihood of purchase.
# 📂 Dataset (Sample)
# Discount (%)ﾠIncome (₹ in thousands)ﾠPurchase
# 5ﾠ20ﾠNo
# 10ﾠ25ﾠNo
# 15ﾠ30ﾠYes
# 20ﾠ35ﾠYes
# 25ﾠ40ﾠYes
# 30ﾠ45ﾠYes
# 35ﾠ50ﾠYes
# 40ﾠ55ﾠYes
# 📝 Tasks for Students
# Load the dataset into Python using pandas.
# Train a Decision Tree Classifier using Discount and Income as features.
# Predict whether a customer with 18% discount and ₹33k income will purchase.
# Visualize the decision tree to understand the rules.



import pandas as pd
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.inspection import DecisionBoundaryDisplay
from sklearn.metrics import accuracy_score, confusion_matrix

# STEP 2 — Upload CSV file

 #uploaded = files.upload()


# STEP 3 — Load dataset

data = pd.read_csv("Customer.csv")


# STEP 4 — Display dataset

print("Dataset:\n")
print(data)


# # 2. Scatter plot with color mapping to target
plt.figure(figsize=(8, 6))
scatter = plt.scatter(data['Discount'], data['Income'], c=data['Purchase'], cmap='viridis', s=100, alpha=0.8)

# Suppose your Series is:
purchase = data['Purchase']   # contains "Yes"/"No"

# Map categories to colors
color_map = {'Yes': 'green', 'No': 'red'}
colors = purchase.map(color_map)

# Use mapped colors in scatter
plt.scatter(data['X'], data['Y'], c=colors)
plt.show()
xlabel="X[0]"
ylabel="X[1]"

# # 3. Add Labels and Legend
plt.colorbar(scatter)
plt.xlabel('Discount (x1)', 'Income (x2)')
plt.ylabel('Purchase')
plt.title('Scatter 2 Inputs vs 1 Target')
plt.grid(True)

X = data[['Discount', 'Income']] # The two input columns are two features
y = data['Purchase']             # The output/label column


# Initialize the model
clf = DecisionTreeClassifier()

# STEP 5 - Train Decision Tree Classifier using Discount and Income as features
clf.fit(X, y)

# Predict whether a customer with 18% discount and ₹33k income will purchase
predictions = clf.predict([[18,33]])
print("Target purchase Predictions:\n")
print(predictions)

# Step 5: Evaluate
y_pred = clf.predict(X)
print("Purchase accuracy score:", accuracy_score(y, y_pred))
print("Confusion Matrix:\n", confusion_matrix(y, y_pred))


# Optional: Visualize the Decision Tree
print("Decision Tree:\n")
plt.figure(figsize=(8,6))
plot_tree(clf, feature_names=['Discount', 'Income'], filled=True)
plt.show()





# h_pad, w_pad, and pad are valid, but the values you used (h_pad=10, w_pad=50, pad=10) are unusually large.
#This will distort spacing. Usually, small values (like 0.5–2) are sufficient.



# # Visualize plot tree

plot_tree(clf,
          feature_names=("X[0]", "X[1]"),
          class_names=("Not purchasing","Purchasing"),
          filled=True,
          rounded=True
         )
plt.show()


# Plot the decision boundary
ax = plt.subplot(9, 2, 9)
plt.tight_layout(h_pad=0.5, w_pad=0.5, pad=0.5)
DecisionBoundaryDisplay.from_estimator(
        clf,
        X,
        cmap=plt.cm.RdYlBu,
        response_method="predict",
        ax=ax,
        xlabel=("X[0]", "X[1]"),
        ylabel=y[0]
)
plt.show()
```
