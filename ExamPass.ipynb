```python
# Scenario: Predicting Exam Pass/Fail Based on Study Hours

# Context
# A bank wants to decide whether to approve a loan application based on the
# applicant’s age. They believe there is a rule-based pattern (not strictly
# linear), so they use a Decision Tree Classifier to model the decision.

from sklearn.tree import DecisionTreeClassifier

X = [[25], [35], [45], [20],[65],[70]]
y = ['No', 'Yes', 'Yes', 'No','No','No']

model = DecisionTreeClassifier()

model.fit(X, y)

print(model.predict([[40]]))
```
