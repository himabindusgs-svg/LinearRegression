```python
# Project:
#Restaurant Review Analyzer
# This is another excellent real-world NLP project for beginners.
#It uses:
# ✅ Tokenization
#✅ Stopwords Removal
#✅ Lemmatization
#✅ Sentiment Analysis
#✅ Named Entity Recognition (NER)
#all together in one practical workflow.
#PROJECT SCENARIO
#A restaurant company receives customer reviews like:
##“I loved the food at Domino's in Delhi.”
#“The service at Pizza Hut was terrible.”
#“Burger King delivery was very fast.”
#The company wants AI to:
#break text into words,
#clean unnecessary words,
#identify root words,
#detect customer sentiment,
#identify important entities like:
#restaurant names,
#locations.

# STEP 1 — Import Libraries

import nltk
import spacy

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
from nltk.sentiment import SentimentIntensityAnalyzer


# STEP 2 — Download Required NLTK Data

nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('vader_lexicon')


# STEP 3 — Load spaCy Model

!python -m spacy download en_core_web_sm

nlp = spacy.load("en_core_web_sm")


# STEP 4 — Input Review

text = "I loved the food at Domino Pizza's in Delhi. The service at Pizza Hut was terrible.The service at Pizza Hut was terrible."


# STEP 5 — Tokenization

tokens = word_tokenize(text)

print("Tokens:\n")

print(tokens)


# STEP 6 — Stopwords Removal

stop_words = set(stopwords.words('english'))

filtered_words = []

for word in tokens:
    if word.lower() not in stop_words:
        filtered_words.append(word)

print("\nAfter Stopwords Removal:\n")

print(filtered_words)


# STEP 7 — Lemmatization

lemmatizer = WordNetLemmatizer()

lemmatized_words = []
for word in filtered_words:
    lemmatized_words.append(
        lemmatizer.lemmatize(word)
    )

print("\nLemmatized Words:\n")

print(lemmatized_words)


# STEP 8 — Sentiment Analysis

sia = SentimentIntensityAnalyzer()

sentiment = sia.polarity_scores(text)

print("\nSentiment Scores:\n")

print(sentiment)


# STEP 9 — Named Entity Recognition

doc = nlp(text)

print("\nRestaurant names and Locations :\n")

for ent in doc.ents:
    print(ent.text, "-", ent.label_)

```
