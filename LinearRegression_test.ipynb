```python
#Context
#An automobile company wants to estimate a car’s fuel efficiency (miles per gallon) based on its engine size (liters). The assumption is that larger engines consume more fuel, so mileage decreases as engine size increases.
#📂 Dataset (Sample)
#Engine Size (Liters)ﾠFuel Efficiency (MPG)
#1.0ﾠ40
#1.2ﾠ38
#1.5ﾠ35
#1.8ﾠ32
#2.0ﾠ30
#2.2ﾠ28
#2.5ﾠ25
#📝 Tasks for Students
#Load the dataset into Python using pandas.
#Plot a scatter graph of Engine Size vs Fuel Efficiency.
#Train a Linear Regression model to predict MPG from engine size.
#Predict the fuel efficiency of a car with a 1.6L engine.




import pandas as pd
import matplotlib.pyplot as plt

from google.colab import files
from sklearn.linear_model import LinearRegression


# STEP 2 — Upload CSV file

# uploaded = files.upload()


# STEP 3 — Load dataset

df = pd.read_csv("sample.csv")


# STEP 4 — Display dataset

print("Dataset:\n")
print(df)


# STEP 5 — Visualize the data

plt.scatter(df['Engine Size (Liters)'], df['Fuel Efficiency (MPG)'])

plt.xlabel("Engine Size (Liters)")
plt.ylabel("Fuel Efficiency (MPG)")

plt.title("Engine Size (Liters) vs Fuel Efficiency (MPG)")

plt.show()


# STEP 6 — Prepare input and output data

X = df[['Engine Size (Liters)']]   # Input
y = df['Fuel Efficiency (MPG)']          # Output


# STEP 7 — Create Linear Regression model

model = LinearRegression()


# STEP 8 — Train the model

model.fit(X, y)


# STEP 9 — Predicted Fuel Efficiency for 10 litres

prediction = model.predict([[10]])

print("\nPredicted Fuel Efficiency for 10 litres:\n")
print(prediction)


# STEP 10 — Draw Regression Line

plt.scatter(df['Engine Size (Liters)'], df['Fuel Efficiency (MPG)'])

plt.plot(df['Engine Size (Liters)'], model.predict(X))

plt.xlabel("Engine Size (Liters)")
plt.ylabel("Fuel Efficiency (MPG)")

plt.title("Linear Regression Line")

plt.show()





```
