```python
import pandas as pd
from sklearn.linear_model import LinearRegression

data = {
    'Advertising Spend': [5, 7, 9, 11, 13],
    'Sales': [50, 70, 90, 110, 130]
}

df = pd.DataFrame(data)

print(df)


X = df[['Advertising Spend']]
y = df['Sales']
model = LinearRegression()

model.fit(X, y)
prediction = model.predict([[10]])

print(prediction)
```
