```python
# Logistic Regression (Different Scenario)
# PROJECT TITLE
# Loan Approval Predictor
# SCENARIO
# A bank receives many loan applications every day.
# Checking each application manually takes time.
# The bank wants AI to predict:
# Loan Approved or Rejected
# based on:
# Monthly Salary

import pandas as pd
from google.colab import files
from sklearn.linear_model import LogisticRegression

# STEP 2 — Upload CSV file

uploaded = files.upload()


# STEP 3 — Load dataset

df = pd.read_csv("Salary.csv")


# STEP 4 — Display dataset

print("Dataset:\n")
print(df)

X=df['Salary']

y=df['Loan_Status']

model=LogisticRegression()

model.fit(X,y)

prediction=model.predict([[150000]])

print(prediction)

print(X.shape)   # should be (n_samples, n_features)

```
